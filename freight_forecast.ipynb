```python
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

def train_lstm_forecast(df, features, target_col, lookback=30):

    # Remove missing values
    data = df[features + [target_col]].dropna().copy()

    # Input and target
    X_raw = data[features].values
    y_raw = data[target_col].values.reshape(-1, 1)

    # Chronological split
    split = int(len(data) * 0.80)

    X_train_raw = X_raw[:split]
    X_test_raw = X_raw[split:]

    y_train_raw = y_raw[:split]
    y_test_raw = y_raw[split:]

    # Scaling
    X_scaler = MinMaxScaler()
    y_scaler = MinMaxScaler()

    X_train_scaled = X_scaler.fit_transform(X_train_raw)
    X_test_scaled = X_scaler.transform(X_test_raw)

    y_train_scaled = y_scaler.fit_transform(y_train_raw)
    y_test_scaled = y_scaler.transform(y_test_raw)

    # -------------------------
    # Create training sequences
    # -------------------------

    X_train = []
    y_train = []

    for i in range(lookback, len(X_train_scaled)):

        X_train.append(
            X_train_scaled[i-lookback:i]
        )

        y_train.append(
            y_train_scaled[i]
        )

    X_train = np.array(X_train)
    y_train = np.array(y_train)

    # -------------------------
    # Create test sequences
    # -------------------------

    X_combined = np.concatenate(
        [X_train_scaled[-lookback:], X_test_scaled],
        axis=0
    )

    y_combined = np.concatenate(
        [y_train_scaled[-lookback:], y_test_scaled],
        axis=0
    )

    X_test = []
    y_test = []

    for i in range(lookback, len(X_combined)):

        X_test.append(
            X_combined[i-lookback:i]
        )

        y_test.append(
            y_combined[i]
        )

    X_test = np.array(X_test)
    y_test = np.array(y_test)

    # -------------------------
    # Build LSTM
    # -------------------------

    model = Sequential()

    model.add(
        LSTM(
            128,
            return_sequences=True,
            input_shape=(
                X_train.shape[1],
                X_train.shape[2]
            )
        )
    )

    model.add(Dropout(0.2))

    model.add(
        LSTM(64)
    )

    model.add(Dropout(0.2))

    model.add(
        Dense(32, activation="relu")
    )

    model.add(
        Dense(1)
    )

    model.compile(
        optimizer="adam",
        loss="mse"
    )

    # -------------------------
    # Early stopping
    # -------------------------

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    # -------------------------
    # Train
    # -------------------------

    history = model.fit(
        X_train,
        y_train,
        epochs=100,
        batch_size=32,
        validation_split=0.1,
        callbacks=[early_stop],
        verbose=0
    )

    # -------------------------
    # Prediction
    # -------------------------

    y_pred_scaled = model.predict(
        X_test,
        verbose=0
    )

    y_pred = y_scaler.inverse_transform(
        y_pred_scaled
    ).flatten()

    y_actual = y_scaler.inverse_transform(
        y_test
    ).flatten()

    return model, y_actual, y_pred, history
```

```python
def recommend_vessel_with_utilization(cargo_quantity):

    vessels = [
        ("Small", 10000),
        ("Handysize", 35000),
        ("Handymax", 55000),
        ("Supramax", 60000),
        ("Panamax", 80000),
        ("Capesize", 200000),
        ("Newcastlemax", 205000),
        ("Chinamax", 400000)
    ]

    for vessel, capacity in vessels:

        utilization = cargo_quantity / capacity

        if 0.75 <= utilization <= 0.95:
            return vessel, capacity, utilization

    # fallback
    for vessel, capacity in vessels:

        if cargo_quantity <= capacity:
            utilization = cargo_quantity / capacity
            return vessel, capacity, utilization

    return "Very Large Bulk Carrier", 400000, cargo_quantity / 400000
```

```python
def calculate_sailing_days(distance_nm, speed_knots):

    sailing_days = distance_nm / (speed_knots * 24)

    return sailing_days
```

```python
def calculate_total_voyage_days(
    sailing_days,
    port_days,
    buffer_days=2
):

    return sailing_days + port_days + buffer_days
```

```python
def calculate_fuel_cost(
    fuel_consumption_tpd,
    voyage_days,
    fuel_price
):

    fuel_quantity = fuel_consumption_tpd * voyage_days

    fuel_cost = fuel_quantity * fuel_price

    return fuel_quantity, fuel_cost
```

```python
def calculate_freight_revenue(
    freight_rate,
    cargo_quantity
):

    return freight_rate * cargo_quantity
```

```python
def calculate_charter_cost(
    freight_rate_usd_day,
    voyage_days
):

    return freight_rate_usd_day * voyage_days
```

```python
def calculate_total_cost(
    charter_cost,
    fuel_cost,
    port_cost=0,
    other_cost=0
):

    return (
        charter_cost
        + fuel_cost
        + port_cost
        + other_cost
    )
```

```python
def voyage_economics(
    vessel_type,
    distance_nm,
    cargo_quantity,
    freight_rate,
    fuel_price,
    speed_knots,
    fuel_consumption_tpd,
    port_days,
    buffer_days=2,
    port_cost=0,
    other_cost=0
):

    # Sailing time
    sailing_days = distance_nm / (speed_knots * 24)

    # Total voyage time
    voyage_days = (
        sailing_days
        + port_days
        + buffer_days
    )

    # Fuel
    fuel_quantity = (
        fuel_consumption_tpd
        * voyage_days
    )

    fuel_cost = (
        fuel_quantity
        * fuel_price
    )

    # Charter
    charter_cost = (
        freight_rate
        * voyage_days
    )

    # Total cost
    total_cost = (
        charter_cost
        + fuel_cost
        + port_cost
        + other_cost
    )

    return {
        "Vessel_Type": vessel_type,
        "Distance_NM": distance_nm,
        "Cargo_Tonnes": cargo_quantity,
        "Sailing_Days": sailing_days,
        "Total_Voyage_Days": voyage_days,
        "Fuel_Quantity_Tonnes": fuel_quantity,
        "Fuel_Cost_USD": fuel_cost,
        "Charter_Cost_USD": charter_cost,
        "Total_Cost_USD": total_cost
    }
```

```python
def compare_vessels(
    cargo_quantity,
    distance_nm,
    freight_rate,
    fuel_price,
    assumptions_df
):

    results = []

    for _, vessel in assumptions_df.iterrows():

        capacity = vessel["Capacity_DWT"]

        # Skip vessels that cannot carry the cargo
        if cargo_quantity > capacity:
            continue

        result = voyage_economics(
            vessel_type=vessel["Vessel_Type"],
            distance_nm=distance_nm,
            cargo_quantity=cargo_quantity,
            freight_rate=freight_rate,
            fuel_price=fuel_price,
            speed_knots=vessel["Speed_knots"],
            fuel_consumption_tpd=vessel["Fuel_Consumption_TPD"],
            port_days=vessel["Port_Days"]
        )

        utilization = (
            cargo_quantity / capacity
        ) * 100

        result["Capacity_DWT"] = capacity
        result["Utilization_%"] = utilization

        results.append(result)

    return pd.DataFrame(results)
```

```python
def calculate_total_risk(
    market_risk,
    fuel_risk,
    port_risk,
    route_risk
):

    total_risk = (
        0.40 * market_risk +
        0.20 * fuel_risk +
        0.20 * port_risk +
        0.20 * route_risk
    )

    return total_risk
```

```python
def risk_category(risk_score):

    if risk_score < 30:
        return "LOW"

    elif risk_score < 60:
        return "MEDIUM"

    elif risk_score < 80:
        return "HIGH"

    else:
        return "VERY HIGH"
```

```python
def opportunity_score(
    forecast_change,
    forecast_confidence,
    utilization,
    total_risk
):

    # Normalize forecast change
    forecast_signal = min(
        100,
        max(0, 50 + forecast_change * 5)
    )

    utilization_score = min(
        100,
        utilization
    )

    score = (
        0.35 * forecast_signal +
        0.25 * forecast_confidence +
        0.20 * utilization_score +
        0.20 * (100 - total_risk)
    )

    return score
```

```python
def final_recommendation(
    opportunity_score,
    forecast_change,
    total_risk,
    forecast_confidence
):

    if (
        opportunity_score >= 75
        and forecast_change > 5
        and total_risk < 60
        and forecast_confidence >= 60
    ):
        return "CHARTER NOW"

    elif (
        opportunity_score >= 50
        and total_risk < 75
    ):
        return "MONITOR"

    else:
        return "WAIT"
```

```python
def black_pearl_risk_engine(
    current_rate,
    forecast_rate,
    forecast_error_std,
    market_risk,
    fuel_risk,
    port_risk,
    route_risk,
    utilization
):

    # Forecast change
    forecast_change = (
        (forecast_rate - current_rate)
        / current_rate
    ) * 100

    # Forecast confidence
    forecast_confidence = (
        calculate_forecast_confidence(
            forecast_rate,
            forecast_error_std
        )
    )

    # Total risk
    total_risk = calculate_total_risk(
        market_risk,
        fuel_risk,
        port_risk,
        route_risk
    )

    # Opportunity
    opportunity = opportunity_score(
        forecast_change,
        forecast_confidence,
        utilization,
        total_risk
    )

    # Recommendation
    recommendation = final_recommendation(
        opportunity,
        forecast_change,
        total_risk,
        forecast_confidence
    )

    return {

        "Current_Rate": current_rate,

        "Forecast_Rate": forecast_rate,

        "Forecast_Change_%": forecast_change,

        "Forecast_Confidence_%":
            forecast_confidence,

        "Market_Risk":
            market_risk,

        "Fuel_Risk":
            fuel_risk,

        "Port_Risk":
            port_risk,

        "Route_Risk":
            route_risk,

        "Total_Risk":
            total_risk,

        "Opportunity_Score":
            opportunity,

        "Recommendation":
            recommendation
    }
```
